```python
from ipywidgets import widgets

# Create the first label
lbl1 = widgets.Label('First number')
# Display the first label
display(lbl1)

# Create the first text input field
text1 = widgets.Text()
# Display the first text input field
display(text1)

# Create the second label
lbl2 = widgets.Label('Second number')
# Display the second label
display(lbl2)

# Create the second text input field
text2 = widgets.Text()
# Display the second text input field
display(text2)

# Create the button
btn = widgets.Button(description="Add")
# Display the button
display(btn)

# Create the result label
result_label = widgets.Label()
# Display the result label
display(result_label)

# Define the function to be called when the button is clicked
def add(b):
    x = int(text1.value)
    y = int(text2.value)
    result_label.value = 'Result = ' + str(x + y)

# Set the function to be called on button click
btn.on_click(add)
```

```python
widgets.Dropdown(
    options=['1', '2', '3'],
    value='2',
    description='Number:',
    disabled=False,
)
```

```python
widgets.Dropdown(
    options=[('One', 1), ('Two', 2), ('Three', 3)],
    value=2,
    description='Number:',
)
```

```python
#5
widgets.ToggleButtons(
    options=['Slow', 'Regular', 'Fast'],
    description='Speed:',
    disabled=False,
    button_style='', # 'success', 'info', 'warning', 'danger' or ''
    tooltips=['Description of slow', 'Description of regular', 'Description of fast'],
#     icons=['check'] * 3
)
```

```python
widgets.IntProgress(
    value=7,
    min=0,
    max=10,
    description='Loading:',
    bar_style='', # 'success', 'info', 'warning', 'danger' or ''
    style={'bar_color': 'maroon'},
    orientation='horizontal'
)
```
